```python
import nltk
```

```python
nltk.download('punkt')
```

```python
from nltk.tokenize import word_tokenize
sentence = "I am learning Natural Language Processing."
tokens = word_tokenize(sentence)
print(tokens)
```

```python
from nltk.tokenize import sent_tokenize
paragraph = "I am learning Natural Language Processing. It is very interesting and useful."
sentences = sent_tokenize(paragraph)
print(sentences)
```

```python
from nltk.tokenize import word_tokenize
sentence="Hardwork is the key to success."
sentence_tokens=word_tokenize(sentence)
print(sentence_tokens)
from nltk.util import bigrams, trigrams, ngrams

sentence_bigrams = list(bigrams(sentence_tokens))
print(sentence_bigrams)

sentence_trigrams = list(trigrams(sentence_tokens))
print(sentence_trigrams)

sentence_ngrams = list(ngrams(sentence_tokens, 4))
print(sentence_ngrams)
```
