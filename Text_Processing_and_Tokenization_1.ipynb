##Task 2: Basic String Operations

```python
import string
import re

# Step 1: Read text from file
with open("sample_text.txt", "r", encoding="utf-8") as f:
    text_data = f.read()

print("Original Text:\n", text_data)
```

```python
# Step 2: Remove leading/trailing spaces
clean_text = text_data.strip()
```

```python
# Step 3: Lowercase
clean_text = clean_text.lower()
```

```python
# Step 4: Remove punctuation
clean_text = clean_text.translate(str.maketrans("", "", string.punctuation))
```

```python
# Step 5: Remove digits
clean_text = re.sub(r"\d+", "", clean_text)
```

```python
# Step 6: Replace multiple spaces with a single space
clean_text = re.sub(r"\s+", " ", clean_text)
```

```python
print("\nCleaned Text:\n", clean_text)
```

##Task 3: Tokenization

```python
import nltk
nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
# Step 1: Read cleaned text from file
with open("sample_text.txt", "r", encoding="utf-8") as f:
    text_data = f.read()
```

```python
# Step 2: Sentence tokenization
sentences = nltk.sent_tokenize(text_data)
print("Sentences:", sentences)
print("Number of Sentences:", len(sentences))
```

```python
# Step 3: Word tokenization
words = nltk.word_tokenize(text_data)
print("Words:", words)
print("Number of Words:", len(words))
```

##Task 4: Word Frequency Analysis

```python
import nltk
from collections import Counter
from nltk.corpus import stopwords

# Download required resources
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
```

```python
# Step 1: Read text
with open("sample_text.txt", "r", encoding="utf-8") as f:
    text_data = f.read()
```

```python
# Step 2: Lowercase and tokenize
tokens = nltk.word_tokenize(text_data.lower())
```

```python
# Step 3: Keep only alphabetic words (remove punctuation/numbers)
tokens = [t for t in tokens if t.isalpha()]
print("Tokens (after cleaning):", tokens)
```

```python
# Step 4: Remove stopwords (Activity)
stop_words = set(stopwords.words("english"))
tokens = [t for t in tokens if t not in stop_words]

print("Tokens (after removing stopwords):", tokens)
```

```python
# Step 5: Frequency count
freq = Counter(tokens)
print("Word Frequency:", freq)
```

```python
# Step 6: Show 5 most common words
print("Most Common Words:", freq.most_common(5))
```

##Task 5: Exporting Results

```python
import pandas as pd

# Converting frequency dictionary to DataFrame
df = pd.DataFrame(freq.items(), columns=["Word", "Frequency"])

print("Word Frequency Table:\n", df)
```

```python
# Save to CSV
df.to_csv("word_frequency.csv", index=False)
print("Word frequency saved to 'word_frequency.csv'")
```
