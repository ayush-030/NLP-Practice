```python
import librosa
import numpy as np
```

# **My (Ayush's) Voice Input**

## Recording of "Hello" word

```python
filename = "/content/drive/MyDrive/NLP Sheets/Hello.wav"
```

```python
y, sr = librosa.load(filename, sr = None)
mfcc = librosa.feature.mfcc(y = y, sr = sr , n_mfcc = 40)
mfcc_mean = np.mean(mfcc, axis = 1)
print("MFCC means values: ", mfcc_mean)
```
